```python
import numpy as np
```

```python
#------------ Sigmoid Function (Activation function) ----------
def sigmoid(x):
    return 1.0/(1.0 + np.exp(-x))
```

```python
#----------- Sigmoid Derivative (Gradient) ---------
def sigmoid_derivative(sigmoid_output):
    return sigmoid_output * (1 - sigmoid_output)
```

```python
#----------- Loss Function ---------------
def mean_squared_error(y, y_hat):
    return 0.5 * (y - y_hat)**2
```

```python
#------------- Feed Forward ---------
def forward_pass(x1, x2, w1, w2, w3):
    z1 = w1*x1 + w2*x2
    h1 = sigmoid(z1)
    z2 = h1*w3
    y_hat = sigmoid(z2)
    return z1, h1, z2, y_hat
```

```python
#---------- Compute Gradients --------
def compute_grad(x_1, x_2, w_3, h_1, yhat, y_):
    dL_dy_hat = yhat - y_
    dL_dz2 = sigmoid_derivative(yhat) * dL_dy_hat
    dL_dw3 = dL_dz2 * h_1
    dL_dz1 = dL_dz2 * w_3 * sigmoid_derivative(h_1)
    dL_dw1 = dL_dz1 * x_1
    dL_dw2 = dL_dz1 * x_2

    return dL_dw1, dL_dw2, dL_dw3


```

```python
x1, x2 = 1.0, 2.0

y = 1.0

w1, w2, w3 =  0.10, -0.30, 0.50

eta = 10

epochs = 1000

for epoch in range(epochs):
     z1, h1, z2, y_hat = forward_pass(x1, x2, w1, w2 ,w3)

     # loss
     error = mean_squared_error(y, y_hat)
     dL_dw1, dL_dw2, dL_dw3 = compute_grad(x1, x2, w3, h1, y_hat, y)

     # Gradient Descent
     w1 -= eta * dL_dw1
     w2 -= eta * dL_dw2
     w3 -= eta * dL_dw3

     print(
        f"epochs: {epoch + 1} "
        f"y_hat = {y_hat:7.3f} "
        f"w1 = {w1}, w2 = {w2}, w3 = {w3} "
        f"error: {error:7.4f}"
        f"x1 {x1}"
        )
```
