```python
import numpy as np
```

```python
#----------- Forward Pass --------------
def forward_pass(x, w1, w2):
    h = x * w1
    y_hat = h * w2
    return h, y_hat
```

```python
#---------- Loss Function: Mean Squared Error ----------
def MSE(y, y_hat):
    return 0.5 * (y - y_hat)**2
```

```python
#-------- Gradients (Backprop) --------
def compute_grads(x, w1, w2, y):
    h, y_hat = forward_pass(x, w1, w2)
    # Loss Derivative
    dL_dy_hat = (y_hat - y)
    # Chain rile
    dL_dw2 = dL_dy_hat * h # since h = x * w1, if u take the deriv with respect to  w2(this bcomes 1)
    dL_dw1 = dL_dy_hat * (x * w2)
    return dL_dw1, dL_dw2
```

```python
#--------- Training Loop -------
def train_net(x, y, w1_init, w2_init, alpha = 0.01, epochs = 10):
    w1, w2 = w1_init, w2_init

    for e in range(epochs):
        h, y_hat = forward_pass(x, w1, w2)
        loss_val = MSE(y, y_hat)

        dL_dw1, dL_dw2 = compute_grads(x, w1, w2, y)

        print(
            f"dL_dw1: {dL_dw1}, dL_dw2: {dL_dw2},\n"
            f"Iteration {e+1:2d} "
            f"w1={w1:7.3f} w2={w2:7.3f} "
            f"y_hat={y_hat:7.3f} "
            f"loss {loss_val:7.3f} "
        )

        w1 = w1 - alpha * dL_dw1
        w2 = w2 - alpha * dL_dw2

    return w1, w2
```

```python
print(train_net(2, 20, 2, 0.5))
```
